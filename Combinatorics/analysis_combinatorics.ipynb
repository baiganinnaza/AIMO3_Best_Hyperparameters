```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import shutil

# 1. Load Data
# Assuming the user ran the experiment and it saved to 'combinatorics_results.csv' or 'experiment_results_algebra.csv'
# Based on file list, we have 'combinatorics_results.csv'
try:
    df = pd.read_csv('combinatorics_results.csv')
    print("Loaded combinatorics_results.csv")
except FileNotFoundError:
    # Fallback if the user named it differently or it hasn't been renamed yet
    if os.path.exists('experiment_results_algebra.csv'):
        df = pd.read_csv('experiment_results_algebra.csv')
        print("Loaded experiment_results_algebra.csv (Combinatorics run)")
    else:
        print("No result file found. Please run the experiment first.")

# Display summary
if 'df' in locals():
    print(f"Total runs: {len(df)}")
    print(f"Overall Success Rate: {df['correct'].mean():.2%}")
    display(df.head())
```

```python
# 2. Analyze Success Rates by Temperature
if 'df' in locals():
    temp_summary = df.groupby('temperature')['correct'].mean().reset_index()
    print("\nSuccess Rate by Temperature:")
    print(temp_summary)

    # Find Best Configuration
    best_config = df[df['correct'] == 1].groupby(['temperature', 'min_p', 'max_p']).size().reset_index(name='success_count')
    # Since we ran 1 trial per config, any correct=1 is a success. 
    # But if we had multiple trials, we'd sort by mean.
    # Let's see which Temperature has the most successes
    print("\nTop performing configurations:")
    print(df[df['correct'] == 1].sort_values(['temperature', 'max_p']))
```

```python
# 3. Visualization
if 'df' in locals():
    unique_temps = sorted(df['temperature'].unique())
    
    for temp in unique_temps:
        temp_df = df[df['temperature'] == temp]
        if temp_df.empty:
            continue
            
        pivot_table = temp_df.pivot(index='max_p', columns='min_p', values='correct')
        
        plt.figure(figsize=(8, 6))
        sns.heatmap(pivot_table, annot=True, cmap="RdYlGn", cbar=False, vmin=0, vmax=1)
        plt.title(f"Correctness Heatmap (Temperature = {temp})")
        plt.xlabel("min_p")
        plt.ylabel("max_p")
        plt.gca().invert_yaxis()
        plt.show()
```

## Conclusion for Combinatorics

Based on the preliminary data (from CSV):
- **Lower Temperatures (0.2)** seem to provide more consistent correct answers (`21818`) for this type of counting problem compared to higher temperatures which might introduce calculations errors or logic drift.
- **Top_P (Max_P)**: High Max_P (0.95+) sometimes leads to failure (`62140` incorrect answer appearing). Conservative sampling seems better.
- **Min_P**: Does not seem to have as drastic an effect as Temperature, but extremely low values combined with high temperatures might be unstable.

**Recommendation:** For Combinatorics tasks involving precise counting, use **Temperature 0.2** with **Max_P 0.9**.

```python
# 4. Project Organization
import os
import shutil

def organize_files():
    folders = {
        'Combinatorics': ['combinatorics_results.csv', 'analysis_combinatorics.ipynb', 'experiment_results_algebra.csv'], 
        'Geometry': ['geometry_results.csv', 'analysis.ipynb'],
        'Number_Theory': ['numbers_results (1).csv', 'analysis_numbers.ipynb', 'results.csv'],
        'Resources': ['AIMO3_Reference_Problems (1).pdf', 'read_pdf.py'],
        'Data': ['reference (1).csv', 'sample_submission (2).csv', 'test.csv']
    }
    
    # Create directories
    for folder in folders:
        if not os.path.exists(folder):
            os.makedirs(folder)
            print(f"Created directory: {folder}")
            
    # Move files
    for folder, filenames in folders.items():
        for filename in filenames:
            if os.path.exists(filename):
                dest = os.path.join(folder, filename)
                # Rename general 'analysis.ipynb' to 'analysis_geometry.ipynb' if moving to Geometry
                if filename == 'analysis.ipynb' and folder == 'Geometry':
                    dest = os.path.join(folder, 'analysis_geometry.ipynb')
                    
                try:
                    shutil.move(filename, dest)
                    print(f"Moved {filename} to {dest}")
                except Exception as e:
                    print(f"Could not move {filename}: {e}")

    print("Organization complete.")

organize_files()
```
